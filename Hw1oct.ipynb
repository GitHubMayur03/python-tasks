Q1: WAP to reverse a given string.

```python
s1 = "facebook"
rev = ""

for ch in s1:
    rev = ch + rev

print(rev)
```

Q2: WAP to check whether string is pallindrome or not.

```python
s = "madam"
rev = ""

for ch in s:
    rev = ch + rev

print(rev)

if s == rev:
    print("Is a Pallindrome.")
else:
     print("Is not a Pallindrome.")
     
```

Q4: WAP to count the number of consonants in a string.

```python
strC = "SpecialOperator"

count1 = 0
count2 = 0

vowels = "aeiou"

for ch in strC.lower():
    if ch in vowels:
        count1 += 1
    elif ch not in vowels:
        count2 +=1
        
print("No of vowels:",count1,", No of consonants:",count2)
```

Q5: WAP to remove all duplicates from a given string.

```python
s = "success"
uniqueCh = ""
for ch in s:
    if ch not in uniqueCh:  
        uniqueCh += ch
print(uniqueCh)
```

Q6: WAP to count the number of uppercase and lowercase characters.

```python
strText = "Hello Python World!"

count_upr = 0
count_lwr = 0

for ch in strText:
    if ch.isupper():    
        count_upr += 1
    elif ch.islower():  
        count_lwr += 1

print("Number of uppercase characters:", count_upr)
print("Number of lowercase characters:", count_lwr)
```
